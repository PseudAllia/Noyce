```python
def all_shuffles(word: str) -> list[str]:

    if len(word) == 1:
        return [word]
    else:
        first_letter = word[0]

        the_rest = word[1:]

        shuffle_the_rest = all_shuffles(the_rest)
        # print(shuffle_the_rest)
        with_char_returned = []

        for new_word in shuffle_the_rest:
            for i in range(len(word)):
                with_char_returned.append(
                        new_word[:i] +
                        first_letter +
                        new_word[i:]
                )

    return list(set(with_char_returned))

```

```python
len(all_shuffles('hanna'))
```

```python
all_shuffles('ab')
```

```python
all_shuffles('sonyaaa')
```
